```python
# !pip install pandas numpy kagglehub scikit-learn
import warnings
warnings.filterwarnings("ignore")
import kagglehub
import os
import shutil
```

```python
dataset_handle = "wordsforthewise/lending-club"
print(f"Downloading dataset: {dataset_handle}...")
downloaded_path = kagglehub.dataset_download(dataset_handle)
print("Downloaded to temporary path:", downloaded_path)

dest_dir = r"data\raw_data"
os.makedirs(dest_dir, exist_ok=True)

for filename in os.listdir(downloaded_path):
    src_file = os.path.join(downloaded_path, filename)
    dest_file = os.path.join(dest_dir, filename)

    if os.path.isfile(src_file):
        shutil.copy2(src_file, dest_file)
        print(f"Copied: {filename}")
    elif os.path.isdir(src_file):
        if os.path.exists(dest_file):
            shutil.rmtree(dest_file)
        shutil.copytree(src_file, dest_file)
        print(f"Copied directory: {filename}")

print(f"\nAll files successfully copied to: {os.path.abspath(dest_dir)}")
```
